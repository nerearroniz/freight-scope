```python
import glob
import os
os.chdir('..')
```

```python
import pandas as pd
import numpy as np
import geopandas as gpd
```

# 1. Import

```python
# Path to the data folder
folder = 'data/raw'

# Define the common
source_name = 'establishments'

# Use glob to find all the filepaths that start with "udalmap"
filepaths = glob.glob(os.path.join(folder, source_name) + '*.csv')
print(filepaths)
```

```python
## Import the establishments datasets into a single DataFrame
df_list = []
for filepath in filepaths:
    path = os.path.abspath(filepath)
    file_name = os.path.basename(filepath) # get only the file name without the path
    year = int(file_name.replace(source_name, '').replace('-', '').replace('.csv', '')) # get the description of the data

    data = pd.read_csv(path, sep=';', decimal=',', index_col=None, encoding='latin-1')
    print(f'The file {path} has been imported')

    data['Year'] = year

    df_list.append(data) # add the processed data to the list
    
# Concatenate all the datasets from udalmap
df_combined = pd.concat(df_list, axis=0, join='inner')
df_combined
```

# 3. Preprocessing

```python
# Identify rows without valid coordinates
df_combined[~pd.to_numeric(df_combined['  coordenada x'], errors = 'coerce').notna() | ~pd.to_numeric(df_combined['  coordenada y'], errors = 'coerce').notna()]
```

```python
# Select the columns of interest
id_vars = ['Código de establecimiento', 'Denominación de establecimiento', 'Municipio', 'CNAE09   ', 'EMPLEST', '  coordenada x', '  coordenada y', 'Year']
value_vars = [col for col in df_combined.columns if col in id_vars]
df_combined_filtered = df_combined[value_vars]
df_combined_filtered = df_combined_filtered.rename(columns = {'Código de establecimiento': 'ID', 'Denominación de establecimiento': 'Name', 'Municipio': 'Municipality', 'CNAE09   ': 'CNAE', 'EMPLEST': 'EMPLEST', '  coordenada x': 'x', '  coordenada y': 'y'})
df_combined_filtered
```

```python
# Drop rows with missing values in critical columns
df_combined_filtered = df_combined_filtered[pd.to_numeric(df_combined_filtered['x'], errors = 'coerce').notna() & pd.to_numeric(df_combined_filtered['y'], errors = 'coerce').notna()]
df_combined_filtered
```

```python
# Change municipality names for consistency
df_combined_filtered['Municipality'] = df_combined_filtered['Municipality'].replace({'Arrasate / Mondragón': 'Arrasate/Mondragón', 'Leintz Gatzaga': 'Leintz-Gatzaga'})
```

# Processing

```python
# ── CNAE classification — González-Feliu et al. (2020) ───────────────────────
_CNAE_RANGES = [
    (111,   322,   "Agriculture"),
    (510,   990,   "Primary and intermediate products"),
    (1011,  1820,  "Primary and intermediate products"),
    (2311,  3320,  "Primary and intermediate products"),
    (3512,  3530,  "Primary and intermediate products"),
    (3600,  3900,  "Primary and intermediate products"),
    (1910,  2229,  "Chemical industry"),
    (4110,  4399,  "Construction industry"),
    (4611,  4690,  "Wholesale"),
    (4711,  4711,  "Food and non-fragile consumer goods"),
    (4721,  4721,  "Food and non-fragile consumer goods"),
    (4723,  4723,  "Food and non-fragile consumer goods"),
    (4725,  4729,  "Food and non-fragile consumer goods"),
    (4776,  4776,  "Food and non-fragile consumer goods"),
    (4722,  4722,  "Butcher's shops"),
    (4724,  4724,  "Bakery retailers"),
    (4741,  4743,  "Hardware stores"),
    (4751,  4759,  "Furnishing shops"),
    (4761,  4765,  "Bookshops"),
    (4771,  4772,  "Clothing, shoes, leather"),
    (4719,  4719,  "Other retail shops"),
    (4730,  4730,  "Other retail shops"),
    (4774,  4775,  "Other retail shops"),
    (4777,  4779,  "Other retail shops"),
    (4791,  4799,  "Other retail shops"),
    (4773,  4773,  "Pharmacies"),
    (4781,  4789,  "Street trading (marketplaces)"),
    (4910,  5122,  "Only Transport"),
    (5210,  5320,  "Transport and warehousing"),
    (5510,  5630,  "Hotels, restaurants, cafés"),
    (4511,  4540,  "Tertiary (non-offices) and services"),
    (5811,  6399,  "Tertiary (non-offices) and services"),
    (6411,  6630,  "Tertiary (non-offices) and services"),
    (6810,  6832,  "Tertiary (non-offices) and services"),
    (8510,  8560,  "Tertiary (non-offices) and services"),
    (8610,  8899,  "Tertiary (non-offices) and services"),
    (9602,  9602,  "Tertiary (non-offices) and services"),
    (6910,  7500,  "Offices"),
    (7711,  8299,  "Offices"),
    (8411,  8430,  "Offices"),
    (9411,  9601,  "Offices"),
    (9603,  9609,  "Offices"),
    (9001,  9329,  "Craftsmen"),
    (9700,  9820,  "Other"),
    (9900,  9900,  "Other"),
]
_GROUP_MAP = {
    "Agriculture":                          "Agriculture",
    "Primary and intermediate products":    "Primary and intermediate products",
    "Chemical industry":                    "Chemical industry",
    "Construction industry":                "Construction industry",
    "Wholesale":                            "Wholesale",
    "Food and non-fragile consumer goods":  "Food and non-fragile retail",
    "Butcher's shops":                      "Food and basic goods retail",
    "Bakery retailers":                     "Food and basic goods retail",
    "Hardware stores":                      "Specialized retail",
    "Furnishing shops":                     "Specialized retail",
    "Bookshops":                            "Specialized retail",
    "Clothing, shoes, leather":             "Specialized retail",
    "Other retail shops":                   "Specialized retail",
    "Pharmacies":                           "Pharmacies",
    "Street trading (marketplaces)":        "Street trading",
    "Only Transport":                       "Transport and warehousing",
    "Transport and warehousing":            "Transport and warehousing",
    "Hotels, restaurants, cafés":           "Hotels, restaurants, cafés",
    "Tertiary (non-offices) and services":  "Tertiary (non-offices) and services",
    "Offices":                              "Offices",
    "Craftsmen":                            "Craftsmen",
    "Other":                                "Other",
}

# EMPLEST ordinal code → employee count range
_EMP_RANGE = {1:    (1, 2),
              2:    (3, 5),
              3:    (6, 9),
              4:    (10, 14),
              5:    (15, 19),
              6:    (20, 49),
              7:    (50, 99),
              8:    (100, 249),
              9:    (250, 499),
              10:   (500, 500)}
```

```python
# Classify CNAE
cnae_ranges = pd.DataFrame(
    _CNAE_RANGES, columns=["min_CNAE", "max_CNAE", "sector"]
    )
conditions = [
        (df_combined_filtered["CNAE"] >= r.min_CNAE) & (df_combined_filtered["CNAE"] <= r.max_CNAE)
        for r in cnae_ranges.itertuples()
    ]
df_combined_filtered.loc[:, "sector"] = np.select(conditions, cnae_ranges["sector"].values, default=None)
df_combined_filtered.loc[:, "group"]  = df_combined_filtered["sector"].map(_GROUP_MAP)
df_combined_filtered
```

```python
# Assign max number of employees
df_combined_filtered.loc[:, "Employees"] = df_combined_filtered["EMPLEST"].map(
    {k: v[1] for k, v in _EMP_RANGE.items()}
)
df_combined_filtered
```

```python
df_combined_filtered.to_csv('data/processed/establishments.csv', index=False, sep=',', decimal='.', encoding='utf-8')
```

```python
# Convert to geospatial data
gpd.GeoDataFrame(
    df_combined_filtered,
    geometry=gpd.points_from_xy(df_combined_filtered["x"], df_combined_filtered["y"]),
    crs='EPSG:25830' # UTM zone 30N — good for Gipuzkoa
    ).to_file('data/processed/establishments.geojson', driver='GeoJSON')
```
